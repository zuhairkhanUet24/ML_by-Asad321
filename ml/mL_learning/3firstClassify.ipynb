# 🤖 Dataset in Machine Learning

## 📚 Introduction

> **datasets are the foundation of Machine learning**
A **dataset** is a collection of data used to **analyze information, find patterns, and train Machine Learning models**.

| 🔧 components | 📌 Purpose |
|:---|:---|
| 📊 **Features** | Input data used by the model | 
| 🎯 **Target** | Output value the model predicts | 
| 📝 **Rows** | Individual data samples | 
| 📋 **Columns** | Data attributes / variables |
