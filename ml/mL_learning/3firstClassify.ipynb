# 🤖 Dataset in Machine Learning

## 📚 Introduction

> **datasets are the foundation of Machine learning**
A **dataset** is a collection of data used to **analyze information, find patterns, and train Machine Learning models**.

| 🔧 components | 📌 Purpose |
|:---|:---|
| 📊 **Features** | Input data used by the model | 
| 🎯 **Target** | Output value the model predicts | 
| 📝 **Rows** | Individual data samples | 
| 📋 **Columns** | Data attributes / variables |

---

## 🔗 ML Workflow

**📥 Data → ⚙️ Processing → 📊 Visualization → 🤖 ML Model**

**load modules**

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import mglearn
from IPython.display import display
from  sklearn.datasets import load_iris
```

```python
iris_dataset = load_iris()
print("Keys of iris_dataset:\n")
# print("keys of iris_dataset:\n{}\n".format(iris_dataset.keys())) instead of this i like to use for loop to print the keys of iris_dataset
for key in iris_dataset.keys():
    print(key,end="  ")
```

**explore keys**

```python
print("\nData:\n{}".format(iris_dataset['data'][:5])) # first five rows of data
```

```python
print("\nTarget names:\n{}".format(iris_dataset['target_names']))
```

```python
print("\ntarget:\n{}".format(iris_dataset['target'])) # first five target values
```

```python
print("\nfeature names:\n{}".format(iris_dataset['feature_names'])  )
```

```python
print("\ndescription:\n{}".format(iris_dataset['DESCR'][:193] + "\n...")) # first five target values
```

<div style="background-color: gray;">
    <span style="color: aqua;">Measuring Success</span>
</div>

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(iris_dataset['data'], iris_dataset['target'], random_state=0)
display(x_train.shape)
display(y_train.shape)
display(x_test.shape)
display(y_test.shape)
```

<div style="color: aqua" >
<h3>data virtualization<h3>
</div>

```python
iris_dataframe = pd.DataFrame(x_train, columns=iris_dataset.feature_names)
pd.plotting.scatter_matrix(iris_dataframe, c=y_train, figsize=(15,15), marker='o', hist_kwds={'bins':20}, s=60, alpha=.8, cmap=mglearn.cm3)
```
